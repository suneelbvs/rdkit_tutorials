# Tutorial 9: Visualisation and Drawing Options

Explore common molecule visualisation utilities that ship with RDKit.


## Objectives

- Generate publication-ready 2D depictions.
- Lay out collections of molecules with grid images and highlights.
- Produce similarity maps to rationalise activity trends.


```python
from rdkit import Chem
from rdkit.Chem import AllChem, Draw
from rdkit.Chem.Draw import SimilarityMaps
```

## Basic Depiction

RDKit draws 2D depictions directly in notebooks. Coordinate generation ensures consistent layouts.


```python
aspirin = Chem.MolFromSmiles('CC(=O)Oc1ccccc1C(=O)O')
AllChem.Compute2DCoords(aspirin)
Draw.MolToImage(aspirin, size=(250, 200), legend='Aspirin')
```

## Grid Images with Highlights

Combine multiple molecules into a single figure and annotate matching substructures.


```python
series = [
    Chem.MolFromSmiles('c1ccncc1'),
    Chem.MolFromSmiles('c1ccccc1O'),
    Chem.MolFromSmiles('c1ccc(Cl)cc1'),
    Chem.MolFromSmiles('c1ccc(N)cc1')
]
patt = Chem.MolFromSmarts('c1ccccc1')
matches = [mol.GetSubstructMatch(patt) for mol in series]
Draw.MolsToGridImage(series, molsPerRow=2, highlightAtomLists=matches, legends=[f"Mol {i+1}" for i in range(len(series))])
```

## Similarity Maps

Similarity maps colour each atom according to its contribution to a fingerprint-based similarity score.


```python
ref = Chem.MolFromSmiles('CCOc1ccccc1O')
probe = Chem.MolFromSmiles('CCOc1ccc(Cl)cc1O')
AllChem.Compute2DCoords(ref)
AllChem.Compute2DCoords(probe)
SimilarityMaps.GetSimilarityMapForFingerprint(ref, probe, SimilarityMaps.GetMorganFingerprint)
```

## Export to SVG

Use the vector backends to create scalable artwork for documentation or presentations.


```python
drawer = Draw.MolDraw2DSVG(300, 200)
drawer.DrawMolecule(aspirin)
drawer.FinishDrawing()
svg = drawer.GetDrawingText()
svg[:200] + '...'
```
